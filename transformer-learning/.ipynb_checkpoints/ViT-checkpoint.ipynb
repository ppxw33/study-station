## Patch Embedding

将图像分割成多个片段和线性 将这些扁平化的斑块投影为一个整体 卷积运算，其中核大小和步幅均为 设置为补丁大小。

````python
class PatchEmbedding(nn.Module):
    def __init__(self, img_size=96, patch_size=16, num_hiddens=512):
        super().__init__()
        def _make_tuple(x):
            if not isinstance(x, (list, tuple)):
                return (x, x)
            return x
        img_size, patch_size = _make_tuple(img_size), _make_tuple(patch_size)
        self.num_patches = (img_size[0] // patch_size[0]) * (
            img_size[1] // patch_size[1])
        self.conv = nn.LazyConv2d(num_hiddens, kernel_size=patch_size,
                                  stride=patch_size)

    def forward(self, X):
        # Output shape: (batch size, no. of patches, no. of channels)
        return self.conv(X).flatten(2).transpose(1, 2)
````

在下例中，输入为 的图像，分别为 的高度和宽度，补丁嵌入输出的补丁线性投影为 长度为 的向量。img_size(img_size//patch_size)**2num_hiddens

## Vision Transformer Encoder

The MLP of the vision Transformer encoder is slightly different from the positionwise FFN of the original Transformer encoder 

视觉 Transformer 编码器中的多层感知机（MLP），与原始 Transformer 编码器的逐位置前馈网络（FFN）存在细微差异。

First, here the activation function uses the Gaussian error linear unit (GELU), which can be considered as a smoother version of the ReLU 

首先，这里使用 ** 高斯误差线性单元（GELU）** 作为激活函数，它可以看作是 ReLU 的平滑版本。

Second, dropout is applied to the output of each fully connected layer in the MLP for regularization.

其次，该多层感知机（MLP）中每个全连接层的输出都会应用 ** dropout** 以实现正则化。

````python
class ViTMLP(nn.Module):
    def __init__(self, mlp_num_hiddens, mlp_num_outputs, dropout=0.5):
        super().__init__()
        self.dense1 = nn.LazyLinear(mlp_num_hiddens)
        self.gelu = nn.GELU()
        self.dropout1 = nn.Dropout(dropout)
        self.dense2 = nn.LazyLinear(mlp_num_outputs)
        self.dropout2 = nn.Dropout(dropout)

    def forward(self, x):
        return self.dropout2(self.dense2(self.dropout1(self.gelu(
            self.dense1(x)))))
````

The vision Transformer encoder block implementation just follows the pre-normalization design in Fig. 11.8.1, where normalization is applied right before multi-head attention or the MLP. In contrast to post-normalization, where normalization is placed right after residual connections, pre-normalization leads to more effective or efficient training for Transformers

视觉 Transformer 编码器模块完全沿用图 11.8.1 中的前置归一化设计，即在多头注意力机制或多层感知机（MLP）执行前进行归一化操作。后置归一化则是在残差连接之后做归一化，与之相比，前置归一化能让 Transformer 模型的训练效果更好、效率更高。

````python
class ViTBlock(nn.Module):
    def __init__(self, num_hiddens, norm_shape, mlp_num_hiddens,
                 num_heads, dropout, use_bias=False):
        super().__init__()
        self.ln1 = nn.LayerNorm(norm_shape)
         # ========== 核心修复处 ==========
        self.attention = d2l.MultiHeadAttention(
            num_hiddens, num_hiddens, num_hiddens,
            num_hiddens, num_heads, dropout, use_bias
        )
#         self.attention = d2l.MultiHeadAttention(num_hiddens, num_heads,
#                                                 dropout, use_bias)

        self.ln2 = nn.LayerNorm(norm_shape)
        self.mlp = ViTMLP(mlp_num_hiddens, num_hiddens, dropout)

    def forward(self, X, valid_lens=None):
        X = X + self.attention(*([self.ln1(X)] * 3), valid_lens)
        return X + self.mlp(self.ln2(X))
````

vision Transformer 编码器模块不会改变输入特征的维度形状。

## Putting It All Together

The forward pass of vision Transformers below is straightforward. First, input images are fed into an PatchEmbedding instance, whose output is concatenated with the “<cls>” token embedding. They are summed with learnable positional embeddings before dropout. Then the output is fed into the Transformer encoder that stacks num_blks instances of the ViTBlock class. Finally, the representation of the “<cls>” token is projected by the network head.
    
下面介绍视觉 Transformer 的前向传播过程，流程十分直观。首先，输入图像送入分块嵌入层，其输出与 ** 类别标记（<cls>）** 的嵌入向量拼接；随后二者与可学习的位置嵌入相加，再执行丢弃操作。接着，结果送入由num_blks个视觉 Transformer 块堆叠而成的 Transformer 编码器。最后，类别标记（<cls>）对应的特征表示经由分类头网络完成映射输出。

````python
# 补全缺失的 Classifier 基类
class Classifier(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        self.num_classes = num_classes

    def accuracy(self, Y_hat, Y, averaged=True):
        Y_hat = Y_hat.reshape((-1, self.num_classes))
        Y = Y.reshape(-1)
        predictions = Y_hat.argmax(dim=1).type(Y.dtype)
        cmp = (predictions == Y).float()
        return cmp.mean() if averaged else cmp

    def loss(self, Y_hat, Y, averaged=True):
        Y_hat = Y_hat.reshape((-1, self.num_classes))
        Y = Y.reshape(-1)
        return nn.functional.cross_entropy(Y_hat, Y, reduction='mean' if averaged else 'none')

# 把补好的类挂载到 d2l.torch，兼容旧写法
d2l.Classifier = Classifier

class ViT(nn.Module):
    def __init__(self, img_size, patch_size, num_hiddens, mlp_num_hiddens,
                 num_heads, num_blks, emb_dropout, blk_dropout, lr=0.1,
                 use_bias=False, num_classes=10):
        super().__init__()
        # 删掉 self.save_hyperparameters()
        # 下面保留你原本的网络层定义
        self.patch_embedding = PatchEmbedding(img_size, patch_size, num_hiddens)
        self.cls_token = nn.Parameter(torch.zeros(1, 1, num_hiddens))
# # 让 ViT 同时继承 d2l.HyperParameters 和 d2l.Classifier
# class ViT(d2l.Classifier, d2l.HyperParameters):
#     def __init__(self, img_size, patch_size, num_hiddens, mlp_num_hiddens,
#                  num_heads, num_blks, emb_dropout, blk_dropout, lr=0.1,
#                  use_bias=False, num_classes=10):
#         super().__init__(num_classes)
# class ViT(d2l.Classifier):
#     """Vision Transformer."""
#     def __init__(self, img_size, patch_size, num_hiddens, mlp_num_hiddens,
#                  num_heads, num_blks, emb_dropout, blk_dropout, lr=0.1,
#                  use_bias=False, num_classes=10):
#         super().__init__(num_classes)

#         self.save_hyperparameters()
        self.patch_embedding = PatchEmbedding(
            img_size, patch_size, num_hiddens)
        self.cls_token = nn.Parameter(torch.zeros(1, 1, num_hiddens))
        num_steps = self.patch_embedding.num_patches + 1  # Add the cls token
        # Positional embeddings are learnable
        self.pos_embedding = nn.Parameter(
            torch.randn(1, num_steps, num_hiddens))
        self.dropout = nn.Dropout(emb_dropout)
        self.blks = nn.Sequential()
        for i in range(num_blks):
            self.blks.add_module(f"{i}", ViTBlock(
                num_hiddens, num_hiddens, mlp_num_hiddens,
                num_heads, blk_dropout, use_bias))
        self.head = nn.Sequential(nn.LayerNorm(num_hiddens),
                                  nn.Linear(num_hiddens, num_classes))

    def forward(self, X):
        X = self.patch_embedding(X)
        X = torch.cat((self.cls_token.expand(X.shape[0], -1, -1), X), 1)
        X = self.dropout(X + self.pos_embedding)
        for blk in self.blks:
            X = blk(X)
        return self.head(X[:, 0])
````

## Training

在 Fashion-MNIST 数据集上训练视觉 Transformer，和第 8 章中卷积神经网络（CNN）的训练方式基本一致。

````python
# 实例化模型
img_size, patch_size = 96, 16
num_hiddens, mlp_num_hiddens, num_heads, num_blks = 512, 2048, 8, 2
emb_dropout, blk_dropout, lr = 0.1, 0.1, 0.1
model = ViT(img_size, patch_size, num_hiddens, mlp_num_hiddens, num_heads,
            num_blks, emb_dropout, blk_dropout, lr)

trainer = d2l.Trainer(max_epochs=10, num_gpus=1)
data = d2l.FashionMNIST(batch_size=128, resize=(img_size, img_size))
trainer.fit(model, data)
````

## Summary and Discussion

对于（Fashion-MNIST）这类小规模数据集，我们实现的视觉 Transformer 性能并未优于8.6 节中的残差网络（ResNet）。即便在拥有 120 万张图像的 ImageNet 数据集上，也能观察到类似现象。这是因为 Transformer 缺少卷积操作具备的优良特性，例如平移不变性与局部关联性。

但当模型规模与训练数据集进一步扩大（如 3 亿张图像）时，情况便会扭转：在图像分类任务中，视觉 Transformer 的表现会大幅超越残差网络，充分体现出 Transformer 在可扩展性上的先天优势（Dosovitskiy 等人，2021）。

视觉 Transformer 的出现，彻底改变了图像数据建模的网络设计格局。随后，借助数据高效训练策略训练得到的 DeiT 模型也证明，视觉 Transformer 在 ImageNet 数据集上同样表现出色（Touvron 等人，2021）。

不过，自注意力机制存在平方级复杂度（详见 11.6 节），这使得 Transformer 架构并不适配超高分辨率图像。为打造计算机视觉领域的通用骨干网络，Swin Transformer 针对图像尺寸带来的平方级计算复杂度问题进行了优化，同时重新引入了类卷积的先验特性。该模型将 Transformer 的应用场景拓展至图像分类之外的各类计算机视觉任务，并取得了当前最优的效果（Liu 等人，2021）。
